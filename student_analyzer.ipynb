```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks  # dictionary: subject -> score

    def validate_marks(self):
        for subject, score in self.marks.items():
            if not isinstance(score, (int, float)):
                raise ValueError(f"Invalid mark for {subject}: must be a number")
            if score < 0 or score > 100:
                raise ValueError(f"Invalid mark for {subject}: must be between 0 and 100")

    def total_marks(self):
        return sum(self.marks.values())

    def average_marks(self):
        if len(self.marks) == 0:
            return 0  # avoid division by zero
        return self.total_marks() / len(self.marks)

    def grade(self):
        avg = self.average_marks()
        if avg >= 90:
            return "A+"
        elif avg >= 75:
            return "A"
        elif avg >= 60:
            return "B"
        elif avg >= 50:
            return "C"
        else:
            return "F"

    def result_summary(self):
        try:
            self.validate_marks()
        except ValueError as e:
            print(f"Error for {self.name}: {e}")
            return

        print(f"--- Result for {self.name} ---")
        for subject, score in self.marks.items():
            print(f"{subject}: {score}")
        print(f"Total Marks: {self.total_marks()}")
        print(f"Average Marks: {self.average_marks():.2f}")
        print(f"Grade: {self.grade()}")
        print("----------------------------\n")


# Example usage
students = [
    Student("Swathi", {"Math": 95, "Science": 88, "English": 92}),
    Student("Rahul", {"Math": 70, "Science": 65, "English": 60}),
    Student("Priya", {"Math": 45, "Science": 50, "English": 40}),
    Student("InvalidCase", {"Math": -10, "Science": "eighty", "English": 110}),  # edge case
]

for s in students:
    s.result_summary()
```

Possible Edge Cases
1. Missing Data
Student has no marks entered.

Some subjects are missing scores.

Name field is empty.

2. Invalid Data Types
Marks entered as strings ("ninety" instead of 90).

Negative marks (-10).

Marks exceeding 100 (120).

3. Zero Subjects
Student has no subjects assigned → division by zero when calculating average.

4. Boundary Values
Marks exactly at grade cutoffs (e.g., 90, 75, 60, 50).

All marks = 0 (should result in F).

All marks = 100 (perfect score, A+).

5. Multiple Students
Large dataset of students (scalability).

Duplicate names (two students named "Rahul").

Students with different subject sets (one has 3 subjects, another has 5).

6. Data Source Issues
CSV file has missing columns.

Extra whitespace or formatting errors.

Inconsistent subject names (Math vs Mathematics).

7. Output Formatting
Very long student names.

Subjects with unusual names (e.g., "Computer Science & Applications").

Ensuring results remain readable
